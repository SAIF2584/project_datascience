```python
df.drop(columns=['Overview','Original_Language','Poster_Url'])
```

```python
def categorize_col(df, col, labels):

    edges = [df[col].describe()['min'],
             df[col].describe()['25%'],
             df[col].describe()['50%'],
             df[col].describe()['75%'],
             df[col].describe()['max']]

    df[col] = pd.cut(df[col], edges, labels=labels, duplicates='drop')

    return df
```

```python
labels = ['not_popular', 'below_avg', 'average', 'popular']

categorize_col(df, 'Vote_Average', labels)

df['Vote_Average'].unique()
```

```python

df['Genre'] = df['Genre'].str.split(', ')


df = df.explode('Genre').reset_index(drop=True)
df.head()
```

```python

#What is the most frequent genre in the dataset?
most_frequent_genre = df['Genre'].value_counts().idxmax()
print(most_frequent_genre)

```

```python

genre_percentages = df['Genre'].value_counts(normalize=True) * 100

for genre, percent in genre_percentages.items():
    print(f"{genre} {percent:.0f}%")
```

```python
#Which year has the most filmmed movies?

most_movies_year = df['Release_Date'].value_counts().idxmax()

print(f"The year with the most filmed movies is: {most_movies_year}")
```

```python
# Calculate the number of unique movies in the highest year (2021)
highest_year = df['Release_Date'].value_counts().idxmax()

# Count unique movie titles for that year
unique_movies_count = df[df['Release_Date'] == highest_year]['Title'].nunique()
print(f"The number of unique movies in the highest year ({highest_year}) is: {unique_movies_count}")
```
